# Does Luxury Sell You a Feeling, Not a Color?
### A linguistic analysis of paint naming conventions, and a generator built from the findings

## Problem Statement
Does the language paint brands use to name their colors signal (or help
construct) a luxury/premium perception — independent of the paint itself?

This project has two parts:
1. **Analysis** — quantify how luxury and mass-market paint brands differ
   in their color-naming conventions, using real catalog data.
2. **Application** — use whatever naming pattern actually distinguishes
   luxury brands to generate new, on-brand color names for any input color.

```python
from wordfreq import word_frequency

def avg_rarity(name):
    words = str(name).lower().split()
    scores = [word_frequency(w, 'en') for w in words]
    return sum(scores) / len(scores) if scores else 0
```

## Findings

| Feature | Luxury (n=425) | Mass-Market (n=6,008) |
|---|---|---|
| Vocabulary rarity (avg score, lower = rarer) | 0.000101 | 0.000212 |
| Contains a real place reference | 5.2% | 2.2% |
| Contains a literal color word | 35.5% | 32.6% |
| Avg word count | 1.79 | 1.84 |

**Luxury brands don't use more literal color words or longer names than
mass-market brands** — those two features showed no reliable pattern.
What *does* hold up consistently: luxury names use **vocabulary roughly
half as common** as mass-market names, and reference **real places over
twice as often**. At the brand level, this forms a clean gradient from
rarest to most common vocabulary: Farrow & Ball (0.000086) → Little
Greene (0.000111) → Sherwin-Williams (0.000155) → Behr (0.000232) — no
brand crosses out of its tier's ordering.

**Data quality note:** Benjamin Moore was excluded from this analysis
after discovering systematic name truncation in the source data (e.g.
the real color "Paper Mache" was recorded as just "Paper"), and after
removing 1,744 duplicate names (44% of the raw file) that would have
badly skewed word-frequency statistics. Place-name detection was
initially attempted with spaCy's NER model, but abandoned after finding
a ~50% false-positive rate on short, context-free product names — a
curated place/nationality word list was used instead.

**Conclusion:** naming sophistication in luxury paint branding comes
from specific, measurable linguistic choices — rare vocabulary and place
references — not from a vague "sounds fancier" effect, and not from
simply avoiding literal color words or writing longer names.

## Application: Luxury Color Name Generator

Since the analysis found that luxury brands specifically lean on **rare
vocabulary** and **real place references** — not longer names or fewer
literal color words — this section builds a generator that applies those
exact patterns: given any color (extracted from an uploaded photo), it
produces new name suggestions styled after real luxury naming
conventions, using a character-level Markov chain trained on the actual
Farrow & Ball / Little Greene dataset.

Every generated name is checked against the real dataset to guarantee
it's genuinely novel, not reproduced from training data.

**Note:** this section uses interactive `ipywidgets` (an upload button
and a generate button). These require running the notebook locally in
Jupyter to use — GitHub's notebook preview only renders a static
snapshot and cannot run the interactive elements.

```python
def broad_family(family):
    if family == 'Neutral':
        return 'Neutral'
    elif family in ['Red', 'Orange/Brown', 'Yellow']:
        return 'Warm'
    else:  # Blue, Green, Purple/Pink
        return 'Cool'

luxury_hex['broad_family'] = luxury_hex['family'].apply(broad_family)
luxury_hex['broad_family'].value_counts()
```

```python
import pandas as pd, re, random, io, colorsys
import ipywidgets as widgets
from PIL import Image

# ---- Load luxury color data directly from your Excel file ----
# IMPORTANT: replace the filename below with your actual file
raw = pd.read_excel(r'C:\Data Analysis Project\paint_color_research_FINAL.xlsx', sheet_name='Luxury', skiprows=3)
raw.columns = ['name', 'code', 'hex', 'brand', 'country', 'tier', 'price']
luxury_hex = raw[['name', 'hex']].dropna().reset_index(drop=True)

# ---- color math ----
def hex_to_hue_sat(hex_code):
    hex_code = str(hex_code).lstrip('#')
    r, g, b = tuple(int(hex_code[i:i+2], 16) for i in (0, 2, 4))
    h, l, s = colorsys.rgb_to_hls(r/255, g/255, b/255)
    return round(h*360, 1), round(s*100, 1)

def hue_to_family(hue, sat):
    if sat < 15: return 'Neutral'
    if hue < 30 or hue >= 330: return 'Red'
    elif hue < 60: return 'Orange/Brown'
    elif hue < 90: return 'Yellow'
    elif hue < 165: return 'Green'
    elif hue < 255: return 'Blue'
    else: return 'Purple/Pink'

def broad_family(hue, sat):
    fam = hue_to_family(hue, sat)
    if fam == 'Neutral': return 'Neutral'
    return 'Warm' if fam in ['Red','Orange/Brown','Yellow'] else 'Cool'

luxury_hex[['hue','sat']] = luxury_hex['hex'].apply(lambda x: pd.Series(hex_to_hue_sat(x)))
luxury_hex['broad_family'] = luxury_hex.apply(lambda r: broad_family(r['hue'], r['sat']), axis=1)

FAMILY_COLOR_WORDS = {
    'Warm': ['red', 'orange', 'rust', 'terracotta', 'amber', 'gold', 'mustard', 'ochre', 'tan'],
    'Cool': ['blue', 'teal', 'navy', 'green', 'sage', 'purple', 'violet', 'plum'],
    'Neutral': ['white', 'grey', 'stone', 'cream', 'taupe'],
}

# ---- Markov chain per broad family ----
def build_markov_chain(names, order=3):
    chain = {}
    for name in names:
        padded = ('~'*order) + str(name).lower() + '~'
        for i in range(len(padded)-order):
            state = padded[i:i+order]
            chain.setdefault(state, []).append(padded[i+order])
    return chain

chains_by_family = {}
for fam in luxury_hex['broad_family'].unique():
    names = luxury_hex[luxury_hex['broad_family']==fam]['name'].tolist()
    chains_by_family[fam] = build_markov_chain(names)

existing_words = set()
for n in luxury_hex['name']:
    existing_words.update(str(n).lower().split())



print(f"Setup complete. Trained on {len(luxury_hex)} luxury names across {len(chains_by_family)} families.")

```

```python
from wordfreq import word_frequency

def generate_word(chain, order=3, max_len=10):
    state = '~'*order
    result = []
    for _ in range(max_len):
        choices = chain.get(state)
        if not choices: break
        nxt = random.choice(choices)
        if nxt == '~' or nxt == ' ':   # stop at end OR at a space - keeps it one word
            break
        result.append(nxt)
        state = (state+nxt)[-order:]
    return ''.join(result)

```

```python
NARROW_COLOR_WORDS = {
    'Red': ['red', 'crimson', 'scarlet', 'rose'],
    'Orange/Brown': ['orange', 'rust', 'terracotta', 'amber', 'tan', 'brown'],
    'Yellow': ['yellow', 'gold', 'mustard', 'ochre'],
    'Green': ['green', 'sage', 'olive', 'jade'],
    'Blue': ['blue', 'teal', 'navy', 'indigo', 'turquoise'],
    'Purple/Pink': ['purple', 'violet', 'plum', 'mauve'],
    'Neutral': ['white', 'grey', 'stone', 'cream', 'taupe'],
}
SAFE_MODIFIERS = {
    'Warm': ['Golden', 'Autumn', 'Warm', 'Deep', 'Soft', 'Rustic', 'Sunset'],
    'Cool': ['Misty', 'Deep', 'Soft', 'Coastal', 'Twilight', 'Quiet'],
    'Neutral': ['Pale', 'Soft', 'Warm', 'Quiet', 'Muted', 'Classic'],
}

```

```python
from wordfreq import word_frequency
def generate_word(chain, order=3, max_len=10):
    state = '~'*order
    result = []
    for _ in range(max_len):
        choices = chain.get(state)
        if not choices: break
        nxt = random.choice(choices)
        if nxt in ('~', ' ', '-', "'"):   # added apostrophe to the stop list
            break
        result.append(nxt)
        state = (state+nxt)[-order:]
    return ''.join(result)
```

```python
existing_full_names = set(str(n).lower().strip() for n in luxury_hex['name'])

def generate_luxury_name(hex_code, n=3, max_attempts=500):
    hue, sat = hex_to_hue_sat(hex_code)
    fam = broad_family(hue, sat)
    narrow_fam = hue_to_family(hue, sat)
    chain = chains_by_family[fam]
    color_word_options = NARROW_COLOR_WORDS[narrow_fam]
    results = []
    seen = set()

    def try_add(candidate):
        key = candidate.lower().strip()
        if key in existing_full_names or key in seen:
            return False
        seen.add(key)
        results.append(candidate)
        return True

    # Guaranteed clean result - now also checked against real names
    attempts = 0
    while len(results) < 1 and attempts < 50:
        attempts += 1
        safe_modifier = random.choice(SAFE_MODIFIERS[fam])
        safe_color = random.choice(color_word_options)
        try_add(f"{safe_modifier} {safe_color.title()}")

    # Creative Markov-generated results
    attempts = 0
    while len(results) < n and attempts < max_attempts:
        attempts += 1
        modifier = generate_word(chain)
        if len(modifier) < 5 or modifier in existing_words:
            continue
        freq = word_frequency(modifier, 'en')
        if freq == 0 or freq > 0.0001:
            continue
        color_word = random.choice(color_word_options)
        try_add(f"{modifier.title()} {color_word.title()}")

    return narrow_fam, results
```

```python
import ipywidgets as widgets
uploader = widgets.FileUpload(accept='image/*', multiple=False)
generate_btn = widgets.Button(description='Generate Names', button_style='success')
output = widgets.Output()

def on_generate_click(b):
    output.clear_output()
    with output:
        if not uploader.value:
            print("Upload a photo first, then click Generate.")
            return
        file_info = list(uploader.value.values())[0] if isinstance(uploader.value, dict) else uploader.value[0]
        img = Image.open(io.BytesIO(file_info['content'])).convert('RGB')
        img_small = img.resize((1,1))
        r, g, b_ = img_small.getpixel((0,0))
        hex_code = '#{:02x}{:02x}{:02x}'.format(r, g, b_)

        fam, names = generate_luxury_name(hex_code)
        print(f"Extracted color: {hex_code}  (family: {fam})\n")
        for name in names:
            print(f"  • {name}")

generate_btn.on_click(on_generate_click)
display(widgets.VBox([uploader, generate_btn, output]))
```
